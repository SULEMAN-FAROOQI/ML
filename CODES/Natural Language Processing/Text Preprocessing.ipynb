INITIAL PREPROCESSING:

```python
# Lowercasing:

data["review"] = data["review"].str.lower() 
print(data.head())
```

```python
# Removing HTML tags:

def remove_html_tags(text): 
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)

data["review"] = data["review"].apply(remove_html_tags)
print(data.head())
```

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)

new_text4 = remove_url(text4)
print(new_text4)
```

```python
# Removing Punctuations:

exclude = string.punctuation
def remove_punc(text):
    return text.translate(str.maketrans('', '', exclude))

data["review"] = data["review"].apply(remove_punc)
print(data.head())
```

```python
# Removing Slangs:

def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words:
            new_text.append(chat_words[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)

data["review"] = data["review"].apply(chat_conversion)
print(data.head())
```

```python
# Spelling Correction:

incorrect_text = 'ceertain conditionas duriing seveal ggenerations aree moodified in the saame maner.'
textBlb = TextBlob(incorrect_text)
print(textBlb.correct().string)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    return " ".join(word for word in text.split() if word not in stop_words)

data["review"] = data["review"].apply(remove_stopwords)
print(data.head())
```

```python
# Handling Emojis: 

def remove_emoji(text):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)

data["review"] = data["review"].apply(remove_emoji)
print(data.sample(5))
```

LEMMATIZATION: 

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

# One-time downloads (the *_tab / *_eng ones are needed by newer NLTK versions)
for pkg in ["punkt", "punkt_tab", "wordnet",
            "averaged_perceptron_tagger", "averaged_perceptron_tagger_eng"]:
    nltk.download(pkg, quiet=True)

lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations = {"?", ":", "!", ".", ",", ";"}

# Build a new list instead of removing while iterating
words = [w for w in nltk.word_tokenize(sentence) if w not in punctuations]

def get_wordnet_pos(tag):
    if tag.startswith("J"):
        return wordnet.ADJ
    if tag.startswith("V"):
        return wordnet.VERB
    if tag.startswith("R"):
        return wordnet.ADV
    return wordnet.NOUN  # default

print("{0:20}{1:20}".format("Word", "Lemma"))
for word, tag in nltk.pos_tag(words):
    print("{0:20}{1:20}".format(word, lemmatizer.lemmatize(word, pos=get_wordnet_pos(tag))))

# Should be used when we need to display output to the user.
```
