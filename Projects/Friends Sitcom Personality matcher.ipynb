```python
import pandas as pd
import numpy as np

data = pd.read_csv("C:\\Users\\sulem\\OneDrive\\Desktop\\Codes\\ML\\Datasets\\Friends.csv")
data.drop("Show", axis = 1, inplace= True)
data.dropna(subset=['Speaker'], inplace=True) # subset=['Speaker'] tells it to only check the Speaker column for missing values, ignore everything else
data.sample(20)
```

```python
data['Speaker'] = data['Speaker'].str.strip().str.title()
# .str.strip() — removes any accidental spaces around the name like "  Monica  " → "Monica"
# .str.title() — capitalizes the first letter of every word, lowercases the rest like "MONICA" → "Monica

name_map = {
    'Rach': 'Rachel', 'Racel': 'Rachel', 'Rahcel': 'Rachel', 'Rache': 'Rachel',
    'Phoe': 'Phoebe', 'Phobe': 'Phoebe',
    'Mnca': 'Monica',
    'Chan': 'Chandler',
    'Estl': 'Estelle',
    'Jnce': 'Janice',
    'Frnz': 'Frank Jr.',
    'Meln': 'Melissa',
    'Angl': 'Angel',
    'Rtst': 'Artist',
    'Flgt': 'Flight Attendant',
    'Mich': 'Michelle',
}
data['Speaker'] = data['Speaker'].replace(name_map)

# Any speaker that starts with these is not a real character
# These are regex patterns. The ^ means "starts with". So r'^\[' means "starts with [".
junk_patterns = [
    r'^\[',          # starts with [
    r'^\(',          # starts with (
    r'^\{',          # starts with {
    r'^Scene',
    r'^Cut To',
    r'^Written By',
    r'^Transcribed By',
    r'^Teleplay',
    r'^Story',
    r'^Produced By',
    r'^Directed By',
    r'^Note',
    r'^Original Airdate',
    r'^Part ',
    r'^Episode',
    r'^Aired',
    r'^Credits',
    r'^Pre-Intro',
    r'^Time Lapse',
]

pattern = '|'.join(junk_patterns) # This joins all patterns into one big pattern using | which means OR in regex.
data = data[~data['Speaker'].str.match(pattern, na=False)]

# data['Speaker'].str.match(pattern) — checks each speaker name, returns True if it matches any junk pattern.
# ~ — means NOT, so we flip True to False and vice versa and keeps only the rows where result is True (meaning NOT junk).

group_keywords = ['And ', ' & ', 'Everyone', 'All ', 'Both', 'Together', 
                  'Gang', 'Girls', 'Guys', 'Boys', 'Others', 'Everybody']

mask = data['Speaker'].apply(lambda x: any(kw in x for kw in group_keywords))
data = data[~mask]

# For each speaker name x, it checks if any of the group keywords exist inside it and ~ keeps only rows where mask is False (single speakers, not groups).

sorted(data['Speaker'].unique())
```

```python
dialogue = data.groupby('Speaker')['Text'].apply(lambda lines: ' '.join(lines)).to_dict()
```

The .groupby('Speaker') — groups all rows by speaker name. Think of it like sorting all lines into separate piles, one pile per character.

Monica pile:  ["Oh my God!", "We were on a break!", "I knew it!"]
Chandler pile: ["Could this BE...", "I'm not great at advice..."]

['Text'] — from each pile, only look at the Text column (we don't need Episode/Season).

.apply(' '.join) — for each character's pile, join all their lines into one single string with a space between them.
["Oh my God!", "We were on a break!", "I knew it!"]
→  "Oh my God! We were on a break! I knew it!"

The .to_dict() — converts the result into a dictionary where the key is the speaker name and the value is all their dialogue.

```python
Data = pd.DataFrame()
Data['character'] = dialogue.keys()
Data['words'] = dialogue.values()
Data['Num of Words'] = Data['words'].apply(lambda x:len(x.split()))

Data = Data.sort_values('Num of Words',ascending=False)
Data = Data.head(50).reset_index(drop=True)

Data.head(20)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(stop_words='english')

embeddings = cv.fit_transform(Data['words']).toarray()
embeddings = embeddings.astype('float64') # For Dimensionality Reduction
embeddings
```

```python
embeddings.shape
```

```python
from sklearn.manifold import TSNE
tsne = TSNE(n_components=2, verbose=1, random_state=33)
z = tsne.fit_transform(embeddings)
```

```python
import plotly.express as px
fig = px.scatter(Data, x = z[: ,0], y = z [:,1], color="character")
fig.show()
```
